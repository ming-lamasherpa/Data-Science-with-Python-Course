# Shifting 

```python
# Moving your data froward or backward 
```

```python
monthly_sales.drop('new_values', axis=1,inplace=True)
monthly_sales
```

```python
monthly_sales['Sales Previous Month'] = monthly_sales['Sales'].shift(1)
monthly_sales
```

```python
plt.figure(figsize=(30,2))
sns.lineplot(x='Order Date', y='Sales', data=monthly_sales)
sns.lineplot(x='Order Date',  y='Sales Previous Month', data=monthly_sales)
```

```python
monthly_sales['monthly_diff']  = monthly_sales['Sales'] - monthly_sales['Sales Previous Month']
```
