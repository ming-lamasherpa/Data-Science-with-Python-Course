```python
import pandas as pd
import numpy as  np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
# dat
data =pd.read_csv(r"C:\Users\Lenovo\Desktop\Power bi files\ship performance\Ship_Performance_Dataset.csv")
data.head(2)
```

```python
# Check Nulll values 
data.isna().sum()
```

```python
# drop all the null values 
data =data.dropna()
data
```

```python
# Select only numericc data type 
df= data.copy() 
```

```python
data = data.select_dtypes(['int', 'float'])
```

```python
data
```

```python
# Scaled the Data 
from  sklearn.preprocessing import StandardScaler 
scaler = StandardScaler()  
```

```python
data_scaled = pd.DataFrame(scaler.fit_transform(data), columns=data.columns)
data_scaled.head(2)
```

```python
# Convert (Decompose) in to 2 D using PCA 
```

```python
from sklearn.decomposition import PCA
```

```python
pca= PCA(n_components = 2) 
pca
```

```python
pca_df = pca.fit_transform(data_scaled)
pca_df
```

```python
pca_df=pd.DataFrame(pca_df, columns=['A', 'B'])
pca_df.head(2)
```

```python
sns.scatterplot(x='A', y='B', data=pca_df)
```

```python
df
```

```python
df['A'] = pca_df['A']
```

```python
df['B'] = pca_df['B']
```

```python
df
```

```python
# Cluster the Data with KMeans and wcss 
from sklearn.cluster import KMeans
```

```python
wcss=[] 
```

```python
for k in range(2,10): 
    model =  KMeans(n_clusters=k) 
    model.fit(pca_df)
    wcss.append(model.inertia_)
```

```python
plt.plot(range(2,10), wcss, marker='D')
```

```python
cluster_range= range(2,10)
sil_score=[] 
```

```python
pca_df.head(1)
```
